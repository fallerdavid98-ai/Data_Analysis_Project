```python
# Importing Libraries
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'],errors="coerce")
```

```python
#No concatenation

df_4th_quarter = df[df["job_posted_date"].dt.quarter==4].copy()

df_4th_quarter["job_posted_date"] = df_4th_quarter["job_posted_date"].dt.month

df_4th_quarter.groupby("job_posted_date")["job_title_short"].count().plot(kind="bar")
```

```python
#With concatenation

df["job_posted_month"] = df["job_posted_date"].dt.strftime("%b")

months = df["job_posted_month"].unique()

month_dict = {month: df[df["job_posted_month"]==month] for month in months}

df_oct = month_dict["Oct"]
df_nov = month_dict["Nov"]
df_dec = month_dict["Dec"]

df_combinded = pd.concat([df_oct,df_nov,df_dec],ignore_index=True)

df_combinded.loc[:,"job_posted_month"].value_counts().plot(kind="bar")
```

```python
df_oct.to_csv("October_Data.csv",index=False)
```
