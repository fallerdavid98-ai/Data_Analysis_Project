```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
Bool_df = df[["job_work_from_home","job_no_degree_mention","job_health_insurance"]].copy()
```

```python
Bool_df["job_work_from_home"]
```

```python
fig, ax = plt.subplots(3,1)

dict_display_names = {
    "job_work_from_home": "Work from Home",
    "job_no_degree_mention": "Job Degree Req.",
    "job_health_insurance": "Health Insurance Offered"
}

#Bool_df["job_work_from_home"].value_counts().plot(kind="pie")

for i,(col, title) in enumerate(dict_display_names.items()):
    print(title)
    ax[i].pie(Bool_df[col].value_counts(),startangle=90,autopct="%1.1f%%",labels=["False","True"])
    ax[i].set_title(title)   
```
