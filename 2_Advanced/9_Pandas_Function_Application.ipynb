```python
# Importing Libraries
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'],errors="coerce")
```

```python
#with public function

df_salary_no_nan = df[pd.notna(df["salary_year_avg"])].copy()

def projected_salary(salary):
    return salary*1.03

df_salary_no_nan["salary_year_inflated"] = df_salary_no_nan["salary_year_avg"].apply(projected_salary)

df_salary_no_nan["inflation_rate"] = df_salary_no_nan["salary_year_inflated"]/df_salary_no_nan["salary_year_avg"]

df_salary_no_nan[["salary_year_inflated","salary_year_avg","inflation_rate"]]
```

```python
#with annonymous function

df_salary_no_nan = df[pd.notna(df["salary_year_avg"])].copy()

df_salary_no_nan["salary_year_inflated"] = df_salary_no_nan["salary_year_avg"].apply(lambda salary: salary*1.03)

df_salary_no_nan["inflation_rate"] = df_salary_no_nan["salary_year_inflated"]/df_salary_no_nan["salary_year_avg"]

df_salary_no_nan[["salary_year_inflated","salary_year_avg","inflation_rate"]]
```

```python
def projected_salary_sr(row):
    if "Senior" in row["job_title_short"]:
        return row["salary_year_avg"]*1.05
    else:
        return row["salary_year_avg"]*1.03
```

```python
df_salary_no_nan2 = df[pd.notna(df["salary_year_avg"])].copy()

df_salary_no_nan2["salary_year_inflated"] = df_salary_no_nan2.apply(projected_salary_sr,axis=1)

df_salary_no_nan2[["job_title_short","salary_year_avg","salary_year_inflated"]]
```
